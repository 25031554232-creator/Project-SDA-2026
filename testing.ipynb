```python
import pandas as pd
import time
from IPython.display import display

df = pd.read_csv("osn.csv")


pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

# =========================================================
# FUNGSI MENAMPILKAN HASIL
# =========================================================

def tampilkan_hasil(dataframe):

    if len(dataframe) > 0:

        dataframe = dataframe.copy()
        dataframe.index = range(1, len(dataframe) + 1)

        print("\nDATA DITEMUKAN")
        print("=" * 120)

        display(dataframe)

        print("=" * 120)
        print("Jumlah Data Ditemukan :", len(dataframe))

    else:
        print("\nData tidak ditemukan.")

# =========================================================
# LINEAR SEARCH PROVINSI
# =========================================================

def linear_search_provinsi(df, keyword):

    hasil = []

    for i in range(len(df)):

        provinsi = str(df.iloc[i]["Provinsi"])

        if keyword.lower() in provinsi.lower():
            hasil.append(df.iloc[i])

    return pd.DataFrame(hasil)

# =========================================================
# LINEAR SEARCH KOTA
# =========================================================

def linear_search_kota(df, keyword):

    hasil = []

    for i in range(len(df)):

        kota = str(df.iloc[i]["Kab/Kota"])

        if keyword.lower() in kota.lower():
            hasil.append(df.iloc[i])

    return pd.DataFrame(hasil)

# =========================================================
# INFORMASI DATA
# =========================================================

print("=" * 100)
print("PROGRAM LINEAR SEARCH DATA OSN")
print("=" * 100)

print("Jumlah Data :", len(df))

# =========================================================
# MENU
# =========================================================

print("\nPILIH JENIS PENCARIAN")
print("1. Cari Berdasarkan Provinsi")
print("2. Cari Berdasarkan Kab/Kota")

pilihan = input("\nMasukkan pilihan (1/2): ")

# =========================================================
# SEARCH PROVINSI
# =========================================================

if pilihan == "1":

    keyword = input("\nMasukkan nama provinsi : ")

    start = time.time()

    hasil = linear_search_provinsi(df, keyword)

    end = time.time()

    tampilkan_hasil(hasil)

    print("\nWaktu Eksekusi :", round(end - start, 6), "detik")

# =========================================================
# SEARCH KOTA
# =========================================================

elif pilihan == "2":

    keyword = input("\nMasukkan nama kota/kabupaten : ")

    start = time.time()

    hasil = linear_search_kota(df, keyword)

    end = time.time()

    tampilkan_hasil(hasil)

    print("\nWaktu Eksekusi :", round(end - start, 6), "detik")

# =========================================================
# INPUT SALAH
# =========================================================

else:
    print("Pilihan tidak valid.")
```

# MENAMPILKAN DATA YANG TERURUT MENGGUNAKAN SHELL SORT & BINARY SEARCH

```python
import pandas as pd
import time
from IPython.display import display

df = pd.read_csv("osn.csv")

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)


def shell_sort(data, kolom):

    n = len(data)
    gap = n // 2

    while gap > 0:

        for i in range(gap, n):

            temp = data[i]

            j = i

            while j >= gap and str(data[j-gap][kolom]).lower() > str(temp[kolom]).lower():

                data[j] = data[j-gap]
                j -= gap

            data[j] = temp

        gap //= 2

    return data


def binary_search(data, kolom, keyword):

    kiri = 0
    kanan = len(data) - 1

    posisi = -1

    keyword = keyword.lower()

    while kiri <= kanan:

        tengah = (kiri + kanan) // 2

        nilai = str(data[tengah][kolom]).lower()

        if nilai == keyword:
            posisi = tengah
            break

        elif nilai < keyword:
            kiri = tengah + 1

        else:
            kanan = tengah - 1

    return posisi


def ambil_semua_data(data, posisi, kolom, keyword):

    hasil = []

    if posisi == -1:
        return pd.DataFrame()

    i = posisi

    while i >= 0 and str(data[i][kolom]).lower() == keyword.lower():
        i -= 1

    i += 1

    while i < len(data) and str(data[i][kolom]).lower() == keyword.lower():

        hasil.append(data[i])

        i += 1

    return pd.DataFrame(hasil)

print("=" * 80)
print("PROGRAM SHELL SORT + BINARY SEARCH")
print("=" * 80)

print("Jumlah Data :", len(df))

print("\n1. Cari Provinsi")
print("2. Cari Kab/Kota")

pilihan = input("\nMasukkan pilihan : ")

if pilihan == "1":

    keyword = input("Masukkan nama provinsi : ")

    data = df.to_dict("records")

    start = time.time()

    data = shell_sort(data, "Provinsi")

    posisi = binary_search(data, "Provinsi", keyword)

    hasil = ambil_semua_data(
        data,
        posisi,
        "Provinsi",
        keyword
    )

    end = time.time()

elif pilihan == "2":

    keyword = input("Masukkan nama kota/kabupaten : ")

    data = df.to_dict("records")

    start = time.time()

    data = shell_sort(data, "Kab/Kota")

    posisi = binary_search(data, "Kab/Kota", keyword)

    hasil = ambil_semua_data(
        data,
        posisi,
        "Kab/Kota",
        keyword
    )

    end = time.time()

else:

    print("Pilihan tidak valid.")
    exit()

if len(hasil) > 0:

    hasil.index = range(1, len(hasil)+1)

    print("\nDATA DITEMUKAN")
    print("=" * 120)

    display(hasil)

    print("=" * 120)
    print("Jumlah Data :", len(hasil))

else:

    print("\nData tidak ditemukan.")

print("\nWaktu Eksekusi :", round(end - start, 6), "detik")
```

# LINEAR SEARCH PROVINSI DAN KAB/KOTA
# MENAMPILKAN DATA DOMINAN

```python
import pandas as pd
import time
from IPython.display import display

df = pd.read_csv("osn.csv")

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)

def tampilkan_hasil(dataframe):

    if len(dataframe) > 0:

        dataframe = dataframe.copy()
        dataframe.index = range(1, len(dataframe) + 1)

        print("\nDATA DITEMUKAN")
        print("=" * 120)

        display(dataframe)

        print("=" * 120)
        print("Jumlah Data Ditemukan :", len(dataframe))

    else:

        print("\nData tidak ditemukan.")

def linear_search_provinsi(df, keyword):

    hasil = []

    for i in range(len(df)):

        provinsi = str(df.iloc[i]["Provinsi"])

        if keyword.lower() in provinsi.lower():

            hasil.append(df.iloc[i])

    return pd.DataFrame(hasil)

def linear_search_kota(df, keyword):

    hasil = []

    for i in range(len(df)):

        kota = str(df.iloc[i]["Kab/Kota"])

        if keyword.lower() in kota.lower():

            hasil.append(df.iloc[i])

    return pd.DataFrame(hasil)

print("=" * 100)
print("PROGRAM LINEAR SEARCH DATA OSN")
print("=" * 100)

print("Jumlah Data :", len(df))
print("Jumlah Provinsi :", df["Provinsi"].nunique())
print("Jumlah Kab/Kota :", df["Kab/Kota"].nunique())

print("\nTOP 10 PROVINSI DENGAN PESERTA TERBANYAK")
print("=" * 100)

provinsi_dominan = df["Provinsi"].value_counts()

for i, (provinsi, jumlah) in enumerate(
        provinsi_dominan.head(10).items(),
        start=1):

    print(f"{i:>2}. {provinsi:<25} : {jumlah} peserta")

print("\nTOP 20 KABUPATEN / KOTA DENGAN PESERTA TERBANYAK")
print("=" * 100)

kota_dominan = df["Kab/Kota"].value_counts()

for i, (kota, jumlah) in enumerate(
        kota_dominan.head(20).items(),
        start=1):

    print(f"{i:>2}. {kota:<35} : {jumlah} peserta")
    
print("\nDAFTAR PROVINSI YANG TERSEDIA")
print("=" * 100)

daftar_provinsi = sorted(df["Provinsi"].dropna().unique())

for i, provinsi in enumerate(daftar_provinsi, start=1):

    print(f"{i:>2}. {provinsi}")

print("\nJumlah Provinsi :", len(daftar_provinsi))

print("\nCONTOH KABUPATEN / KOTA YANG TERSEDIA")
print("=" * 100)

daftar_kota = sorted(df["Kab/Kota"].dropna().unique())

for i, kota in enumerate(daftar_kota[:20], start=1):

    print(f"{i:>2}. {kota}")

print("\nTotal Kab/Kota Dalam Dataset :", len(daftar_kota))

print("\n" + "=" * 100)
print("MENU PENCARIAN")
print("=" * 100)
print("1. Cari Berdasarkan Provinsi")
print("2. Cari Berdasarkan Kab/Kota")

pilihan = input("\nMasukkan pilihan (1/2): ")

if pilihan == "1":

    keyword = input("\nMasukkan nama provinsi : ")

    start = time.time()

    hasil = linear_search_provinsi(df, keyword)

    end = time.time()

    tampilkan_hasil(hasil)

    print("\nWaktu Eksekusi :", round(end - start, 6), "detik")


elif pilihan == "2":

    keyword = input("\nMasukkan nama kota/kabupaten : ")

    start = time.time()

    hasil = linear_search_kota(df, keyword)

    end = time.time()

    tampilkan_hasil(hasil)

    print("\nWaktu Eksekusi :", round(end - start, 6), "detik")


else:

    print("\nPilihan tidak valid.")
```
